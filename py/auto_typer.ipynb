```python
MAX_TEXT_REPEAT = 1
MESSAGE_INTERVAL = 0   
DELAY_BEFORE_START = 5  

print(f"Focus on input. Starting in {DELAY_BEFORE_START} seconds...")
time.sleep(DELAY_BEFORE_START)
```

```python
MAX_INDEX = len(WORDS) - 1
index = 0
can_repeat = MAX_TEXT_REPEAT > 1

try:
    while True:
        # Random word repeated 2–MAX_REPEAT times
        message = WORDS[index]
        
        if can_repeat:
            repeat_count = random.randint(1, MAX_TEXT_REPEAT)
            message = " ".join([message] * repeat_count)

        # message = message.strip() 

        if message != "":     
            pyautogui.typewrite(message,_pause=False)
            # time.sleep(0.1) 
            pyautogui.press("enter")     
            time.sleep(MESSAGE_INTERVAL)
        if index < MAX_INDEX:
            index +=1
        elif index >= MAX_INDEX:
            index = 0


except KeyboardInterrupt:
    print("\nStopped by user.")
```
