```python
data = []
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)
```

```python
def filterDataByType(dataList,type):
    return [x for x in dataList.get(type,[]) if x.get('type') == type]
```

```python
url = "https://node.parkingticket.in/api/test/getOnlineUsers"

# url = "http://82.112.238.169:1100/api/test/getOnlineUsers"

res = req.get(url)

knownTypes = ["stand","display","customer"]

type = "stand"
# type = "display"
# type = "customer"



if(res.status_code == 200):
    data = res.json()
    if "parkingticket.in" in url:
        onlineUsers = ["onlineUsers"]
        totalOnline = 0
        for t in knownTypes:
            c = len(filterDataByType(data,t))
            totalOnline+=c
            onlineUsers.append(f"{t} : {c}")
        onlineUsers.append(f"Total : {totalOnline}")
        print("\n".join(onlineUsers))
        data = filterDataByType(data,type)

```

```python

df = pd.DataFrame(data)

if len(data) > 0 :
    df['connectedOn'] = pd.to_datetime(df['connectedOn'], format='mixed')
    df = df.sort_values(by='connectedOn',ascending=False)

 
df
```
